**Project Goal:**

Investigate how agriculture performance varies Across season and identify meaningful patterns, trends,relationships and differences in available data

**1. Problem Statement:**

* Agricultural activities are influenced by seasonal variations in environmental conditions,
farming practices, resource availability and market conditions. As a result, agricultural
performance may differ from one season to another.

* However, raw agricultural data does not clearly explain how agricultural performance changes
across seasons or what patterns can be observed in different seasonal conditions.

* The problem is to analyze the given agricultural dataset and investigate seasonal differences
in agricultural performance by identifying meaningful patterns, trends, relationships and
variations within the available data.

### Distribution of Crops

```python
crop_counts = df['Crop'].value_counts()

plt.figure(figsize=(8, 8))
plt.pie(crop_counts, labels=crop_counts.index, autopct='%1.1f%%', startangle=140, colors=sns.color_palette('pastel'))
plt.title('Distribution of Crops')
plt.axis('equal') # Equal aspect ratio ensures that pie is drawn as a circle.
plt.show()
```

### Distribution of Seasons

```python
season_counts = df['Season'].value_counts()

plt.figure(figsize=(8, 8))
plt.pie(season_counts, labels=season_counts.index, autopct='%1.1f%%', startangle=140, colors=sns.color_palette('pastel'))
plt.title('Distribution of Seasons')
plt.axis('equal') # Equal aspect ratio ensures that pie is drawn as a circle.
plt.show()
```

```python
plt.figure(figsize=(12, 7))
sns.countplot(data=df, x='Season', hue='Crop', palette='viridis')
plt.title('Distribution of Crops Across Seasons')
plt.xlabel('Season')
plt.ylabel('Number of Farms (or occurrences)')
plt.xticks(rotation=45, ha='right')
plt.legend(title='Crop', bbox_to_anchor=(1.05, 1), loc='upper left')
plt.tight_layout()
plt.show()
```

**Farms Per State**

```python
state_farm_counts = df.groupby('State')['Farm_ID'].count().reset_index()
state_farm_counts = state_farm_counts.sort_values(by='Farm_ID', ascending=False)

plt.figure(figsize=(12, 6))
ax = sns.barplot(x='State', y='Farm_ID', data=state_farm_counts, palette='viridis', hue='State', legend=False)
plt.title('Number of Farms per State')
plt.xlabel('State')
plt.ylabel('Number of Farms')
plt.xticks(rotation=45, ha='right') # Rotate state names for better readability

# Add values on top of the bars
for p in ax.patches:
    ax.annotate(f'{p.get_height():.0f}',
                (p.get_x() + p.get_width() / 2., p.get_height()),
                ha='center', va='center',
                xytext=(0, 9),
                textcoords='offset points')

plt.tight_layout()
plt.show()
```

### Average Profit per Crop Type

### Water Used vs. Water Efficiency

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv('/content/seasonal_agriculture_performance_dataset.csv')

plt.figure(figsize=(10, 6))
sns.scatterplot(x='Water_Used_m3', y='Water_Efficiency_t_per_1000m3', data=df)
plt.title('Water Used vs. Water Efficiency')
plt.xlabel('Water Used (m³)')
plt.ylabel('Water Efficiency (tonnes/1000m³)')
plt.grid(True, linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```

### Production of Crops by State

```python
production_by_state_crop = df.groupby(['State', 'Crop'])['Production_Tonnes'].sum().reset_index()

plt.figure(figsize=(16, 8))
ax = sns.barplot(x='State', y='Production_Tonnes', hue='Crop', data=production_by_state_crop, palette='tab10')
plt.title('Production of Different Crops by State')
plt.xlabel('State')
plt.ylabel('Production (Tonnes)')
plt.xticks(rotation=45, ha='right')
plt.legend(title='Crop', bbox_to_anchor=(1.05, 1), loc='upper left')
plt.grid(axis='y', linestyle='--', alpha=0.7);

# Add values on top of the bars, rotated vertically
for p in ax.patches:
    ax.annotate(f'{p.get_height():.0f}',
                (p.get_x() + p.get_width() / 2., p.get_height()),
                ha='center', va='bottom', # 'bottom' aligns text to the top of the bar
                xytext=(0, 5), # Slight vertical offset
                textcoords='offset points',
                fontsize=9,
                color='black',
                rotation=90) # Rotate text by 90 degrees

plt.tight_layout()
plt.show()
```

### Multivariate Analysis: Pairplot of Key Numerical Variables

```python
selected_variables = [
    'Yield_Tonnes_Ha',
    'Rainfall_mm',
    'Avg_Temperature_C',
    'Water_Used_m3',
    'Profit_INR'
]

sns.pairplot(df[selected_variables], diag_kind='kde')
plt.suptitle('Pairwise Relationships of Key Agricultural Performance Variables', y=1.02)
plt.tight_layout()
plt.show()
```

### Correlation Matrix of Numerical Variables

```python
import numpy as np

# Select only numerical columns for correlation calculation
numerical_df = df.select_dtypes(include=np.number)

# Calculate the correlation matrix
correlation_matrix = numerical_df.corr()

plt.figure(figsize=(20, 18))
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', fmt='.2f', linewidths=.5)
plt.title('Correlation Matrix of Numerical Variables')
plt.tight_layout()
plt.show()
```
